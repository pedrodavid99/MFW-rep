## I WANT TO SAVE THE DATA AS A PARKET OBJET SO I CAN STORE FASTER SO I NEED TO PASS EVERY COL TO ONE TYPE

```python
def columns_with_multiple_types(df):
    cols = []
    for col in df.columns:
        types = set(type(x) for x in df[col] if pd.notnull(x))
        if len(types) > 1:
            cols.append((col, types))
    return cols

# Example usage:
# result = columns_with_multiple_types(your_dataframe)
# print(result)
```

```python
columns_with_multiple_types(lapses)
```

```python
def convert_columns_to_int_strict(df, cols):
    for col in cols:
        # Replace common non-numeric placeholders with NaN
        df[col] = df[col].replace(['#N/D', 'N/A', 'NA', 'nan', 'NaN', ''], pd.NA)
        df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')
    return df

# Example usage:
# df = convert_columns_to_int_strict(df, ['col1', 'col2'])
```

```python
cols_with_nan = [col for col in lapses_pp.columns if lapses_pp[col].isna().any()]
print(cols_with_nan)
```

```python
# 1. Get the mode for each column by SEXO_tom
modes = lapses_pp.groupby('SEXO_tom')[['new_estatura_tom', 'PESO_tom', 'IDADE_tom', 'IMC_tom']].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)

# 2. Get the possible classes for SEXO_tom
sexo_classes = lapses_pp['SEXO_tom'].dropna().unique()

# 3. Find rows where SEXO_tom is NaN
mask = lapses_pp['SEXO_tom'].isna()

# 4. For each such row, randomly assign a class and fill the other columns with the corresponding mode
for idx in lapses_pp[mask].index:
    random_class = np.random.choice(sexo_classes)
    lapses_pp.at[idx, 'SEXO_tom'] = random_class
    for col in ['new_estatura_tom', 'PESO_tom', 'IDADE_tom', 'IMC_tom']:
        lapses_pp.at[idx, col] = modes.loc[random_class, col]
```

```python
# For each column, fill NaN values with the mode of the corresponding SEXO_tom class
for col in ['new_estatura_tom', 'PESO_tom', 'IDADE_tom', 'IMC_tom']:
    lapses_pp[col] = lapses_pp.groupby('SEXO_tom')[col].transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else np.nan))
```

```python
lapses_pp.to_parquet("lapses_pp_1.parquet", index=False)
```
